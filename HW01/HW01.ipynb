````python
res = zip(cities, amounts_int, paid)
print(list(res))
orders = []
for c,a,p in zip(cities, amounts_int, paid):
    order = {
        'city':c,
        'amount':a,
        'paid':p
    }
    print(order)
    orders.append(order)
````

9.Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью while посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
start_amount = 15000
course_amount = 400000
i = 0
while start_amount <=course_amount:
    i+=1
    start_amount+=15000
print(i)
````
