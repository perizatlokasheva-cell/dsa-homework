3.Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.

````python
category_amount = {}
for r in rows:
    if r["status"] == "оплачен":
        amount = r["amount"].strip().replace(" ","").replace(",",".")
        cat = r["category"]
        if cat in category_amount:
            category_amount[cat] +=float(amount)
        else:
            category_amount[cat] = float(amount)
print(category_amount)
````

````python
category_count
````

````python
category_mean = {}
for (c,cnt), (a,amnt) in zip(category_count.items(), category_amount.items()):
    if c== a:
        category_mean[c] = round(amnt/cnt)
print(category_mean)
````

````python
for c in category_count:
    print(f"{c:>12}  {category_amount[c]:>10}  {category_mean[c]:>10}")
````

8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
amount_month = {}
for r in rows:
    order_date = r["order_date"]
    order_date = order_date[:7]
    amount = round(float(r["amount"].strip().replace(" ","").replace(",",".")))
    if r["status"] == 'оплачен':
        if order_date in amount_month:
            amount_month[order_date] += amount
        else:
            amount_month[order_date] = amount
print(amount_month)
        
    
````

````python
for a in amount_month:
    if a == "2026-09":
        amount_sep = amount_month[a]
        print(amount_sep)
    elif a == "2026-10":
        amount_oct = amount_month[a]
        print(amount_oct)

print(100-round(100*amount_sep/amount_oct) )
````

Изменение октября к сентябрю в процентах = 29% в сторону уменьшения

6.Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.

````python
cities = {}
cities_unpaid = {}
for r in rows:
    city = r["city"]
    cln_city = clean_city(city)
    cities[cln_city] = cities.get(cln_city,0) + 1
    if r["status"] != 'оплачен':
        cities_unpaid[cln_city] = cities_unpaid.get(cln_city,0) + 1
        
print(cities)
print(cities_unpaid)
````
